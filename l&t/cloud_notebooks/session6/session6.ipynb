```python
synthetic_rng = np.random.default_rng(42)
n_samples = 100

synthetic_age = synthetic_rng.integers(20, 61, size=n_samples)
synthetic_rating = synthetic_rng.uniform(2.5, 5.0, size=n_samples)
synthetic_salary = (
    30_000
    + 900 * synthetic_age
    + 5_000 * synthetic_rating
    + synthetic_rng.normal(0, 5_000, size=n_samples)
)

synthetic_df = pd.DataFrame({
    "Age": synthetic_age,
    "Rating": synthetic_rating,
    "Salary": synthetic_salary,
})

# Add missing values to the input features
for synthetic_col in ["Age", "Rating"]:
    synthetic_missing_rows = synthetic_rng.choice(n_samples, size=8, replace=False)
    synthetic_df.loc[synthetic_missing_rows, synthetic_col] = np.nan

# Fill missing feature values with each column's median
synthetic_df[["Age", "Rating"]] = synthetic_df[["Age", "Rating"]].fillna(
    synthetic_df[["Age", "Rating"]].median()
)

display(synthetic_df)

synthetic_model = LinearRegression()
synthetic_model.fit(synthetic_df[["Age", "Rating"]], synthetic_df["Salary"])
```
