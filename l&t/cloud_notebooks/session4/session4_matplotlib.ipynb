```python
import matplotlib.pyplot as plt

students = ["Aman", "Riya", "Kabir", "Neha"]
marks = [72, 45, 56, 81]

plt.bar(students,marks)
plt.title("Students Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temprature = [28, 30, 31, 29, 31]

plt.plot(days, temprature, marker="o")
plt.title("Temprature Trend")
plt.xlabel("Days")
plt.ylabel("Temprature")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
languages = ["Python", "Java", "C++", "JavaScript"]
students = [45, 30, 25, 35] 
plt.bar(languages, students) 
plt.title("Programming Language Preference") 
plt.xlabel("Language") 
plt.ylabel("Number of Students") 
plt.show()
```

```python
import matplotlib.pyplot as plt 
activities = ["Study", "Sleep", "College", "Other"] 
hours = [6, 8, 7, 3] 
plt.pie(hours, labels=activities, autopct="%1.1f%%") 
plt.title("A Student's Day") 
plt.show()
```

```python
import matplotlib.pyplot as plt 
study_hours = [1, 2, 3, 4, 5, 6] 
marks = [45, 52, 60, 68, 79, 88] 
plt.scatter(study_hours, marks) 
plt.title("Study Hours vs Marks") 
plt.xlabel("Study Hours") 
plt.ylabel("Marks") 
plt.show()
```
