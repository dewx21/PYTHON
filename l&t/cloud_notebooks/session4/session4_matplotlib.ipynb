```python
import matplotlib.pyplot as plt

students = ["Aman", "Riya", "Kabir", "Neha"]
marks = [72, 45, 56, 81]

plt.bar(students,marks)
plt.title("Students Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```
