```python
import numpy as np
data=np.array([10,20,30,40,50])

print("Array:", data)
print("Mean:", np.mean(data))
print("Maximum:", np.max(data))
print("Double Values:", data * 2)
```

```python
import pandas as pd

# Convert the existing NumPy array into a pandas Series.
# Use Series when working with a single labeled column of data.
numbers = pd.Series(data, name="Values")

# Display the Series.
print("Series:")
print(numbers)

# Create a DataFrame from the Series.
# Use DataFrame when working with table-like data containing rows and columns.
df = pd.DataFrame({"Values": data})

print("\nDataFrame:")
print(df)

# View the first rows of the DataFrame.
print("\nFirst rows:")
print(df.head())

# Get information about columns, data types, and memory usage.
print("\nDataFrame information:")
df.info()

# Calculate common descriptive statistics.
print("\nSummary statistics:")
print(df.describe())

# Select one column.
print("\nValues column:")
print(df["Values"])

# Filter rows using a condition.
print("\nValues greater than 25:")
print(df[df["Values"] > 25])

# Add a new calculated column.
df["Double"] = df["Values"] * 2

print("\nDataFrame with a new column:")
print(df)

# Save the DataFrame as a CSV file.
# This is useful for storing processed data.
df.to_csv("numbers.csv", index=False)
```
