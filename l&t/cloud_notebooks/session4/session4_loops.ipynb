```python
# Functions and loop examples

# Basic function
def greet(name="World"):
    return f"Hello, {name}!"

print(greet())
print(greet("Python"))


# Positional and keyword arguments
def describe(name, age, *, city="Unknown"):
    return f"{name}, {age}, {city}"

print(describe("Alex", 25, city="London"))


# *args and **kwargs
def summarize(*numbers, **details):
    return {
        "total": sum(numbers),
        "count": len(numbers),
        "details": details,
    }

print(summarize(1, 2, 3, subject="Math"))


# Lambda function
square = lambda x: x * x
print(square(5))


# Recursive function
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))


# Function with documentation
def add(a, b):
    """Return the sum of two values."""
    return a + b

print(add(2, 3))
print(add.__doc__)


# for loop over a list
names = ["Ada", "Grace", "Linus"]

for name in names:
    print(name)


# for loop with range
for number in range(1, 6):
    print(number)


# range with step
for number in range(10, 0, -2):
    print(number)


# enumerate
for index, name in enumerate(names, start=1):
    print(index, name)


# zip
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(name, score)


# Loop with else
for number in range(3):
    print(number)
else:
    print("Loop completed")


# break
for number in range(10):
    if number == 5:
        break
    print(number)


# continue
for number in range(10):
    if number % 2 == 0:
        continue
    print("Odd:", number)


# pass
for number in range(3):
    if number == 1:
        pass
    print(number)


# while loop
counter = 0

while counter < 5:
    print(counter)
    counter += 1


# while with break
counter = 0

while True:
    counter += 1
    if counter == 3:
        break
    print(counter)


# while with continue
counter = 0

while counter < 5:
    counter += 1
    if counter == 3:
        continue
    print(counter)


# Nested loops
for row in range(1, 4):
    for column in range(1, 4):
        print(row, column)


# List comprehension
squares = [number ** 2 for number in range(10)]
print(squares)


# Comprehension with condition
even_squares = [
    number ** 2
    for number in range(10)
    if number % 2 == 0
]
print(even_squares)


# Dictionary comprehension
square_map = {number: number ** 2 for number in range(5)}
print(square_map)


# Set comprehension
unique_remainders = {number % 3 for number in range(10)}
print(unique_remainders)


# Generator function
def countdown(start):
    while start > 0:
        yield start
        start -= 1

for value in countdown(5):
    print(value)


# map, filter, and reduce
from functools import reduce

values = [1, 2, 3, 4, 5]

print(list(map(square, values)))
print(list(filter(lambda x: x % 2 == 0, values)))
print(reduce(lambda a, b: a + b, values))


# Exception handling in a loop
for value in ["10", "abc", "20"]:
    try:
        print(int(value))
    except ValueError:
        print("Invalid number:", value)
    finally:
        print("Finished processing:", value)
```
