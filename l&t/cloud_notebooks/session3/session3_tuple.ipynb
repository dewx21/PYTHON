```python
from collections import namedtuple

# ============================================================
# Beginner's Tutorial: Python Tuples
# ============================================================

# A tuple is an ordered collection of values.
# Tuples are immutable, which means their contents cannot be changed
# after the tuple has been created.

# ------------------------------------------------------------
# 1. Creating Tuples
# ------------------------------------------------------------

# An empty tuple
empty_tuple = ()

# A tuple containing several values
numbers = (10, 20, 30, 40)

# Tuples can contain different data types
person = ("Alice", 25, "Engineer")

# Parentheses are optional, but usually recommended for readability
colors = "red", "green", "blue"

# A one-item tuple requires a trailing comma
single_item = (42,)

# Without the comma, this is only an integer
not_a_tuple = (42)

print(empty_tuple)
print(numbers)
print(person)
print(type(single_item))
print(type(not_a_tuple))


# ------------------------------------------------------------
# 2. Accessing Tuple Items
# ------------------------------------------------------------

fruits = ("apple", "banana", "cherry", "orange")

# Indexing starts at 0
print(fruits[0])  # apple
print(fruits[1])  # banana

# Negative indexes access items from the end
print(fruits[-1])  # orange
print(fruits[-2])  # cherry


# ------------------------------------------------------------
# 3. Slicing Tuples
# ------------------------------------------------------------

# Syntax: tuple[start:stop:step]
# The stop index is not included.

print(fruits[1:3])   # ('banana', 'cherry')
print(fruits[:2])    # First two items
print(fruits[2:])    # From index 2 to the end
print(fruits[:])     # Copy of the entire tuple
print(fruits[::2])   # Every second item
print(fruits[::-1])  # Reverse the tuple


# ------------------------------------------------------------
# 4. Tuple Immutability
# ------------------------------------------------------------

coordinates = (10, 20)

# The following line would cause a TypeError:
# coordinates[0] = 99

# Although tuple items cannot be replaced, a tuple can contain
# mutable objects such as lists.
data = ("Python", [1, 2, 3])

# The list inside the tuple can still be modified
data[1].append(4)
print(data)


# ------------------------------------------------------------
# 5. Tuple Packing
# ------------------------------------------------------------

# Packing means placing multiple values into one tuple.
packed_values = 100, 200, 300

print(packed_values)
print(type(packed_values))


# ------------------------------------------------------------
# 6. Tuple Unpacking
# ------------------------------------------------------------

# Unpacking assigns tuple values to separate variables.
student = ("John", 21, "Computer Science")

name, age, subject = student

print(name)
print(age)
print(subject)

# The number of variables must normally match the number of items.


# ------------------------------------------------------------
# 7. Extended Unpacking with *
# ------------------------------------------------------------

numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers

print(first)   # 1
print(middle)  # [2, 3, 4]
print(last)    # 5

# The starred variable collects remaining values into a list.
first, *remaining = numbers
print(first)
print(remaining)


# ------------------------------------------------------------
# 8. Swapping Variables Using Tuples
# ------------------------------------------------------------

a = 10
b = 20

# Python packs the right side and unpacks it into a and b.
a, b = b, a

print(a)  # 20
print(b)  # 10


# ------------------------------------------------------------
# 9. Tuple Methods
# ------------------------------------------------------------

values = (1, 2, 3, 2, 4, 2)

# count() returns how many times a value appears
print(values.count(2))  # 3

# index() returns the first position of a value
print(values.index(3))  # 2

# index() can search within a range
print(values.index(2, 2))  # Search from index 2


# ------------------------------------------------------------
# 10. Useful Built-in Functions
# ------------------------------------------------------------

numbers = (5, 10, 15, 20)

print(len(numbers))   # Number of items
print(min(numbers))   # Smallest item
print(max(numbers))   # Largest item
print(sum(numbers))   # Total
print(sorted(numbers))  # Returns a sorted list
print(any(numbers))   # True if at least one item is truthy
print(all(numbers))   # True if every item is truthy


# ------------------------------------------------------------
# 11. Checking Membership
# ------------------------------------------------------------

languages = ("Python", "Java", "C++")

print("Python" in languages)
print("Ruby" not in languages)


# ------------------------------------------------------------
# 12. Looping Through a Tuple
# ------------------------------------------------------------

for language in languages:
    print(language)

# enumerate() provides both the index and the value
for index, language in enumerate(languages):
    print(index, language)


# ------------------------------------------------------------
# 13. Nested Tuples
# ------------------------------------------------------------

matrix = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)

print(matrix[0])       # First row
print(matrix[1][2])    # Row 2, column 3

for row in matrix:
    for value in row:
        print(value)


# ------------------------------------------------------------
# 14. Converting Between Tuples and Other Collections
# ------------------------------------------------------------

# List to tuple
numbers_list = [1, 2, 3]
numbers_tuple = tuple(numbers_list)
print(numbers_tuple)

# Tuple to list
colors_tuple = ("red", "green", "blue")
colors_list = list(colors_tuple)
print(colors_list)

# String to tuple
letters = tuple("Python")
print(letters)

# Range to tuple
range_tuple = tuple(range(5))
print(range_tuple)


# ------------------------------------------------------------
# 15. Adding Values to a Tuple
# ------------------------------------------------------------

# Tuples cannot be changed directly.
# However, a new tuple can be created by joining tuples.

old_tuple = (1, 2, 3)
new_tuple = old_tuple + (4, 5)

print(new_tuple)

# Repeating a tuple
repeated_tuple = ("Hi",) * 3
print(repeated_tuple)


# ------------------------------------------------------------
# 16. Removing Values from a Tuple
# ------------------------------------------------------------

# Tuples do not have remove() or pop() methods.
# Convert the tuple to a list, modify it, then convert it back.

items = ("pen", "book", "bag")

temporary_list = list(items)
temporary_list.remove("book")
items = tuple(temporary_list)

print(items)


# ------------------------------------------------------------
# 17. Sorting Tuples
# ------------------------------------------------------------

scores = (40, 10, 30, 20)

# sorted() returns a list
sorted_scores = sorted(scores)
print(sorted_scores)

# Convert the sorted list back into a tuple
sorted_scores_tuple = tuple(sorted_scores)
print(sorted_scores_tuple)

# Sort in descending order
print(tuple(sorted(scores, reverse=True)))


# ------------------------------------------------------------
# 18. Comparing Tuples
# ------------------------------------------------------------

# Tuples are compared item by item.
print((1, 2) == (1, 2))
print((1, 2) < (2, 1))

# A tuple with fewer matching items is considered smaller
print((1, 2) < (1, 3))


# ------------------------------------------------------------
# 19. Using zip() to Create Tuples
# ------------------------------------------------------------

names = ("Alice", "Bob", "Charlie")
ages = (25, 30, 35)

# zip() combines corresponding values.
combined = tuple(zip(names, ages))

print(combined)

for name, age in combined:
    print(f"{name} is {age} years old.")


# ------------------------------------------------------------
# 20. Returning Multiple Values from a Function
# ------------------------------------------------------------

def calculate(a, b):
    """Return multiple values packed inside a tuple."""
    total = a + b
    difference = a - b
    product = a * b
    return total, difference, product


result = calculate(10, 5)
print(result)

# Unpack the returned tuple
total, difference, product = calculate(10, 5)

print(total)
print(difference)
print(product)


# ------------------------------------------------------------
# 21. Tuple as a Dictionary Key
# ------------------------------------------------------------

# Tuples can be dictionary keys if all their contents are immutable.
locations = {
    (40.7128, -74.0060): "New York",
    (34.0522, -118.2437): "Los Angeles"
}

print(locations[(40.7128, -74.0060)])


# ------------------------------------------------------------
# 22. Named Tuples
# ------------------------------------------------------------


# A named tuple allows access by both index and attribute name.
Book = namedtuple("Book", ["title", "author", "year"])

book = Book("Python Basics", "Alex Smith", 2024)

print(book.title)
print(book.author)
print(book[2])


# ============================================================
# Summary
# ============================================================

# Important tuple facts:
#
# 1. Tuples are ordered.
# 2. Tuples use parentheses or commas.
# 3. Tuples are immutable.
# 4. A one-item tuple requires a trailing comma.
# 5. Tuples support indexing and slicing.
# 6. Tuples can be packed and unpacked.
# 7. Useful methods include count() and index().
# 8. Tuples can contain different data types.
# 9. Tuples can be nested.
# 10. Tuples are useful for fixed collections of values.
```
