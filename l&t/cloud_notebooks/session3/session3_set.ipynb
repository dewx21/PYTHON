```python
# Python Sets: Step-by-Step Guide
# A set is an unordered collection of unique, hashable values.

# ------------------------------------------------------------
# 1. Creating sets
# ------------------------------------------------------------

empty_set = set()          # Correct way to create an empty set
numbers = {1, 2, 3, 4}
mixed_set = {1, "Python", 3.14, True}

print(numbers)

# Duplicate values are automatically removed
duplicates = {1, 2, 2, 3, 3, 3}
print(duplicates)          # {1, 2, 3}

# Convert another iterable into a set
values = [1, 2, 2, 3, 4]
unique_values = set(values)
print(unique_values)

# Strings are converted into a set of unique characters
letters = set("banana")
print(letters)


# ------------------------------------------------------------
# 2. Adding elements
# ------------------------------------------------------------

numbers.add(5)             # Adds one element
print(numbers)

numbers.update([6, 7, 8])  # Adds multiple elements
print(numbers)

numbers.update({9, 10}, [11, 12])
print(numbers)


# ------------------------------------------------------------
# 3. Removing elements
# ------------------------------------------------------------

numbers.remove(12)         # Removes an element
                           # Raises KeyError if the element is absent

numbers.discard(100)       # Removes an element if present
                           # Does not raise an error if absent

removed_item = numbers.pop()  # Removes and returns an arbitrary element
print("Removed:", removed_item)

print(numbers)

# Remove all elements
temporary_set = {1, 2, 3}
temporary_set.clear()
print(temporary_set)       # set()


# ------------------------------------------------------------
# 4. Checking membership and size
# ------------------------------------------------------------

colors = {"red", "green", "blue"}

print("red" in colors)         # True
print("yellow" not in colors)  # True
print(len(colors))             # Number of elements


# ------------------------------------------------------------
# 5. Copying sets
# ------------------------------------------------------------

original = {1, 2, 3}
shallow_copy = original.copy()

print(shallow_copy)

# Changes to the copy do not affect the original
shallow_copy.add(4)
print(original)
print(shallow_copy)


# ------------------------------------------------------------
# 6. Set union
# Combines all unique elements from two or more sets
# ------------------------------------------------------------

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

union_result = set_a.union(set_b)
print(union_result)

# The | operator also performs union
print(set_a | set_b)

# union() can accept multiple iterables
print(set_a.union(set_b, {7, 8}))

# update() changes the original set
set_a.update(set_b)
print(set_a)


# ------------------------------------------------------------
# 7. Set intersection
# Finds elements common to all sets
# ------------------------------------------------------------

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

intersection_result = set_a.intersection(set_b)
print(intersection_result)

# The & operator also performs intersection
print(set_a & set_b)

# Intersection with multiple sets
print(set_a.intersection(set_b, {4, 5, 6}))

# intersection_update() changes the original set
set_a.intersection_update(set_b)
print(set_a)


# ------------------------------------------------------------
# 8. Set difference
# Finds elements in the first set but not in the second
# ------------------------------------------------------------

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

difference_result = set_a.difference(set_b)
print(difference_result)

# The - operator also performs difference
print(set_a - set_b)

# difference_update() changes the original set
set_a.difference_update(set_b)
print(set_a)


# ------------------------------------------------------------
# 9. Symmetric difference
# Finds elements that exist in either set, but not both
# ------------------------------------------------------------

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

symmetric_result = set_a.symmetric_difference(set_b)
print(symmetric_result)

# The ^ operator also performs symmetric difference
print(set_a ^ set_b)

# symmetric_difference_update() changes the original set
set_a.symmetric_difference_update(set_b)
print(set_a)


# ------------------------------------------------------------
# 10. Comparing sets
# ------------------------------------------------------------

small_set = {1, 2}
large_set = {1, 2, 3, 4}

# issubset(): all elements are contained in another set
print(small_set.issubset(large_set))
print(small_set <= large_set)

# Proper subset: subset but not equal
print(small_set < large_set)

# issuperset(): contains all elements of another set
print(large_set.issuperset(small_set))
print(large_set >= small_set)

# Proper superset: superset but not equal
print(large_set > small_set)

# isdisjoint(): no common elements
set_x = {1, 2}
set_y = {3, 4}
print(set_x.isdisjoint(set_y))


# ------------------------------------------------------------
# 11. Equality and inequality
# ------------------------------------------------------------

first = {1, 2, 3}
second = {3, 2, 1}
third = {1, 2, 4}

print(first == second)     # True; order does not matter
print(first != third)      # True


# ------------------------------------------------------------
# 12. Iterating through a set
# ------------------------------------------------------------

languages = {"Python", "Java", "C++"}

for language in languages:
    print(language)

# Sets are unordered, so use sorted() for predictable display
for language in sorted(languages):
    print(language)


# ------------------------------------------------------------
# 13. Set comprehension
# ------------------------------------------------------------

squares = {number ** 2 for number in range(1, 6)}
print(squares)

even_numbers = {number for number in range(20) if number % 2 == 0}
print(even_numbers)


# ------------------------------------------------------------
# 14. Useful built-in functions
# ------------------------------------------------------------

data = {10, 20, 30, 40}

print(len(data))        # Number of elements
print(min(data))        # Smallest element
print(max(data))        # Largest element
print(sum(data))        # Sum of numeric elements
print(sorted(data))     # Returns a sorted list
print(list(data))       # Converts the set to a list
print(tuple(data))      # Converts the set to a tuple


# ------------------------------------------------------------
# 15. Immutable sets with frozenset
# ------------------------------------------------------------

fixed_set = frozenset([1, 2, 3])

print(fixed_set)

# frozenset supports set operations
print(fixed_set.union({3, 4}))
print(fixed_set.intersection({2, 3, 4}))

# This would raise AttributeError because frozensets are immutable:
# fixed_set.add(4)


# ------------------------------------------------------------
# 16. Important rules
# ------------------------------------------------------------

# Set elements must be hashable.
valid_set = {1, "text", (2, 3)}

# Lists and dictionaries cannot be set elements:
# invalid_set = {[1, 2]}       # TypeError
# invalid_set = {{"a": 1}}     # TypeError

# Sets do not support indexing:
# print(numbers[0])            # TypeError

# Use membership testing instead:
print(3 in {1, 2, 3})


# ------------------------------------------------------------
# 17. Practical example: remove duplicate values
# ------------------------------------------------------------

names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
unique_names = set(names)

print(unique_names)


# ------------------------------------------------------------
# 18. Practical example: common and unique items
# ------------------------------------------------------------

students_python = {"Alice", "Bob", "Charlie"}
students_java = {"Bob", "Charlie", "David"}

print("All students:", students_python | students_java)
print("Both courses:", students_python & students_java)
print("Only Python:", students_python - students_java)
print("Only one course:", students_python ^ students_java)
```
