```python
"""
Python dictionaries: a complete practical guide.

A dictionary stores data as key-value pairs:

    {key: value}

Keys must be hashable and unique. Values can be any Python object.
Dictionaries preserve insertion order in modern Python versions.
"""

# ============================================================
# 1. Creating dictionaries
# ============================================================

empty_dictionary = {}
also_empty = dict()

student = {
    "name": "Ada",
    "age": 36,
    "language": "Python",
}

# Keys may be strings, numbers, tuples, or other hashable objects.
mixed_keys = {
    "text": "value",
    42: "number key",
    (1, 2): "tuple key",
}

# Values may have any type.
varied_values = {
    "integer": 10,
    "float": 3.14,
    "list": [1, 2, 3],
    "nested": {"city": "London"},
    "function": len,
}

# Creating a dictionary from keyword arguments.
person = dict(name="Grace", age=28, profession="programmer")

# Creating a dictionary from an iterable of key-value pairs.
coordinates = dict([
    ("x", 10),
    ("y", 20),
])

# Creating a dictionary with the same default value for several keys.
default_scores = dict.fromkeys(["math", "science", "history"], 0)
print(default_scores)


# ============================================================
# 2. Dictionary keys and values
# ============================================================

# Keys must be unique. A later value replaces an earlier value.
duplicate_keys = {
    "color": "red",
    "color": "blue",
}
print(duplicate_keys)  # {'color': 'blue'}

# Valid hashable keys include strings, numbers, tuples, and booleans.
valid_keys = {
    "name": "Ada",
    1: "one",
    3.14: "pi",
    True: "true",
    (1, 2): "tuple",
}

# Lists, sets, and dictionaries cannot be keys because they are mutable.
# invalid = {[1, 2]: "list key"}  # TypeError

# Values can be mutable.
data = {
    "numbers": [1, 2, 3],
    "settings": {"dark_mode": True},
}


# ============================================================
# 3. Reading values
# ============================================================

user = {
    "name": "Ada",
    "email": "ada@example.com",
    "age": 36,
}

# Access a value using square brackets.
print(user["name"])

# Missing keys raise KeyError.
# print(user["phone"])  # KeyError

# get() safely returns None, or a specified default.
print(user.get("phone"))
print(user.get("phone", "Not provided"))

# The second argument to get() is only returned when the key is missing.
print(user.get("age", 0))  # 36

# Check whether a key exists.
print("email" in user)
print("phone" not in user)

# Membership checks keys, not values.
print("Ada" in user)  # False
print("Ada" in user.values())  # True


# ============================================================
# 4. Adding and changing items
# ============================================================

profile = {}

# Add a new key-value pair.
profile["name"] = "Grace"

# Add another item.
profile["role"] = "engineer"

# Change an existing value.
profile["role"] = "scientist"

# update() adds new items and replaces existing items.
profile.update({
    "country": "United Kingdom",
    "role": "computer scientist",
})

# update() also accepts keyword arguments.
profile.update(language="Python", active=True)

# update() can accept another dictionary.
profile.update({"projects": 4})

# update() can accept an iterable of pairs.
profile.update([
    ("education", "mathematics"),
    ("verified", True),
])

print(profile)


# ============================================================
# 5. Removing items
# ============================================================

inventory = {
    "apples": 10,
    "bananas": 5,
    "oranges": 8,
}

# del removes an item. A missing key raises KeyError.
del inventory["bananas"]
print(inventory)

# pop() removes and returns a value.
apples_count = inventory.pop("apples")
print(apples_count)
print(inventory)

# pop() can use a default for a missing key.
missing_item = inventory.pop("mangoes", 0)
print(missing_item)

# popitem() removes and returns the newest key-value pair.
last_item = inventory.popitem()
print(last_item)

# clear() removes every item.
inventory.clear()
print(inventory)

# del can remove the entire dictionary variable.
temporary = {"a": 1}
del temporary
# print(temporary)  # NameError


# ============================================================
# 6. Inspecting dictionaries
# ============================================================

book = {
    "title": "Python",
    "pages": 300,
    "published": 2024,
}

# len() returns the number of key-value pairs.
print(len(book))

# keys() returns a dynamic view of keys.
print(book.keys())

# values() returns a dynamic view of values.
print(book.values())

# items() returns a dynamic view of key-value pairs.
print(book.items())

# Convert views to lists when a list is needed.
print(list(book.keys()))
print(list(book.values()))
print(list(book.items()))

# A view changes when the original dictionary changes.
keys_view = book.keys()
book["author"] = "Example Author"
print(keys_view)


# ============================================================
# 7. Iterating through dictionaries
# ============================================================

scores = {
    "Alice": 90,
    "Bob": 82,
    "Charlie": 95,
}

# Iterating directly produces keys.
for name in scores:
    print(name)

# Iterate over keys explicitly.
for name in scores.keys():
    print(name)

# Iterate over values.
for score in scores.values():
    print(score)

# Iterate over both keys and values.
for name, score in scores.items():
    print(f"{name}: {score}")

# Enumerate dictionary items with an index.
for position, (name, score) in enumerate(scores.items(), start=1):
    print(position, name, score)


# ============================================================
# 8. Dictionary unpacking
# ============================================================

first = {"a": 1, "b": 2}
second = {"c": 3, "d": 4}

# Unpack keys into a list.
print([*first])

# Merge dictionaries using **.
merged = {**first, **second}
print(merged)

# If keys overlap, the later dictionary wins.
left = {"color": "red", "size": "large"}
right = {"color": "blue"}

print({**left, **right})


# ============================================================
# 9. Modern dictionary merge operators
# ============================================================

one = {"a": 1, "b": 2}
two = {"b": 20, "c": 3}

# | creates a new merged dictionary.
combined = one | two
print(combined)

# |= updates the left dictionary in place.
one |= two
print(one)


# ============================================================
# 10. Dictionary comprehensions
# ============================================================

# Basic dictionary comprehension.
squares = {number: number ** 2 for number in range(1, 6)}
print(squares)

# Transform an existing dictionary.
prices = {
    "book": 10,
    "pen": 2,
    "bag": 25,
}

discounted_prices = {
    item: price * 0.9
    for item, price in prices.items()
}
print(discounted_prices)

# Add a condition.
expensive_items = {
    item: price
    for item, price in prices.items()
    if price > 5
}
print(expensive_items)

# Swap keys and values.
original = {"a": 1, "b": 2}
swapped = {value: key for key, value in original.items()}
print(swapped)

# Create a dictionary from two lists.
names = ["Ada", "Grace", "Alan"]
ages = [36, 28, 42]
people = {name: age for name, age in zip(names, ages)}
print(people)


# ============================================================
# 11. Nested dictionaries
# ============================================================

company = {
    "name": "Example Corp",
    "address": {
        "city": "London",
        "country": "UK",
    },
    "employees": {
        "employee_1": {
            "name": "Ada",
            "role": "developer",
        },
        "employee_2": {
            "name": "Grace",
            "role": "engineer",
        },
    },
}

# Access nested values.
print(company["address"]["city"])
print(company["employees"]["employee_1"]["name"])

# Change a nested value.
company["address"]["city"] = "Cambridge"

# Add a nested value.
company["address"]["postal_code"] = "CB1"

print(company)


# ============================================================
# 12. setdefault()
# ============================================================

# setdefault() returns the existing value if the key exists.
settings = {"theme": "dark"}
current_theme = settings.setdefault("theme", "light")
print(current_theme)
print(settings)

# If the key is missing, setdefault() inserts the default value.
language = settings.setdefault("language", "English")
print(language)
print(settings)

# Useful for grouping values.
groups = {}

for name, department in [
    ("Ada", "research"),
    ("Grace", "research"),
    ("Alan", "engineering"),
]:
    groups.setdefault(department, []).append(name)

print(groups)


# ============================================================
# 13. copy() and shallow copies
# ============================================================

original_data = {
    "name": "Ada",
    "skills": ["Python", "Mathematics"],
}

# Assignment creates another reference to the same dictionary.
same_data = original_data
same_data["name"] = "Changed"
print(original_data["name"])  # Also changed

# copy() creates a shallow copy.
original_data["name"] = "Ada"
shallow_copy = original_data.copy()
shallow_copy["name"] = "Grace"
print(original_data["name"])  # Unchanged

# Nested mutable objects are still shared in a shallow copy.
shallow_copy["skills"].append("Algorithms")
print(original_data["skills"])  # Also contains "Algorithms"

# deepcopy() recursively copies nested objects.
import copy

deep_copy = copy.deepcopy(original_data)
deep_copy["skills"].append("Databases")
print(original_data["skills"])
print(deep_copy["skills"])


# ============================================================
# 14. Equality and identity
# ============================================================

dictionary_a = {"x": 1, "y": 2}
dictionary_b = {"y": 2, "x": 1}
dictionary_c = dictionary_a

# Dictionary equality depends on key-value contents, not insertion order.
print(dictionary_a == dictionary_b)  # True

# is checks whether two variables reference the same object.
print(dictionary_a is dictionary_b)  # False
print(dictionary_a is dictionary_c)  # True


# ============================================================
# 15. Sorting dictionaries
# ============================================================

scores = {
    "Alice": 82,
    "Bob": 95,
    "Charlie": 88,
}

# sorted(dictionary) sorts keys by default.
print(sorted(scores))

# Sort items by key.
by_name = dict(sorted(scores.items()))
print(by_name)

# Sort items by value.
by_score = dict(sorted(scores.items(), key=lambda item: item[1]))
print(by_score)

# Sort values from highest to lowest.
highest_first = dict(
    sorted(scores.items(), key=lambda item: item[1], reverse=True)
)
print(highest_first)


# ============================================================
# 16. Dictionary methods summary
# ============================================================

method_dictionary = {"a": 1, "b": 2}

# Return a shallow copy.
print(method_dictionary.copy())

# Return a value, or a default if the key is absent.
print(method_dictionary.get("a"))
print(method_dictionary.get("missing", 0))

# Return a dynamic view of keys.
print(method_dictionary.keys())

# Remove and return a key-value pair.
print(method_dictionary.pop("a"))

# Remove and return the newest key-value pair.
print(method_dictionary.popitem())

# Insert a default value if a key is absent.
method_dictionary.setdefault("new", 100)
print(method_dictionary)

# Return a dynamic view of values.
print(method_dictionary.values())

# Return a dynamic view of key-value pairs.
print(method_dictionary.items())

# Remove all items.
method_dictionary.clear()
print(method_dictionary)

# Create a dictionary from keys and one shared default value.
print(dict.fromkeys(["x", "y"], 0))

# Add or replace items.
method_dictionary.update({"one": 1, "two": 2})
print(method_dictionary)


# ============================================================
# 17. Built-in functions commonly used with dictionaries
# ============================================================

numbers = {"a": 10, "b": 20, "c": 30}

print(len(numbers))            # Number of pairs
print(list(numbers))           # List of keys
print(tuple(numbers))          # Tuple of keys
print(set(numbers))            # Set of keys
print(any(numbers.values()))   # True if any value is truthy
print(all(numbers.values()))   # True if all values are truthy
print(min(numbers.values()))   # Smallest value
print(max(numbers.values()))   # Largest value
print(sum(numbers.values()))   # Sum of numeric values


# ============================================================
# 18. Truthiness
# ============================================================

# An empty dictionary is false in a Boolean context.
empty = {}
nonempty = {"value": 1}

print(bool(empty))     # False
print(bool(nonempty))  # True

if nonempty:
    print("The dictionary contains items.")

if not empty:
    print("The dictionary is empty.")


# ============================================================
# 19. Safe access patterns
# ============================================================

configuration = {
    "database": {
        "host": "localhost",
        "port": 5432,
    }
}

# Safe access for one level.
host = configuration.get("database", {}).get("host", "unknown")
print(host)

# Avoid assuming a key exists.
if "database" in configuration:
    print(configuration["database"])

# Catch KeyError when direct access is required.
try:
    print(configuration["missing"])
except KeyError:
    print("The key does not exist.")


# ============================================================
# 20. Dictionary keys must be hashable
# ============================================================

# Immutable objects can generally be dictionary keys.
hashable_dictionary = {
    "text": 1,
    10: 2,
    (1, 2): 3,
    frozenset({1, 2}): 4,
}

# Mutable objects cannot be keys.
# bad_dictionary = {[1, 2]: "not allowed"}  # TypeError


# ============================================================
# 21. Using dictionaries for counting
# ============================================================

words = ["python", "code", "python", "data", "code", "python"]
word_counts = {}

for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)


# ============================================================
# 22. Using dictionaries for lookup tables
# ============================================================

number_names = {
    1: "one",
    2: "two",
    3: "three",
}

number = 2
print(number_names.get(number, "unknown"))


# ============================================================
# 23. A dictionary subclass and __missing__()
# ============================================================

class ZeroDictionary(dict):
    """Return 0 for missing keys without inserting them."""

    def __missing__(self, key):
        return 0


counts = ZeroDictionary()
counts["apples"] += 1
counts["apples"] += 1
counts["oranges"] += 1

print(counts)
print(counts["missing"])  # Returns 0
print("missing" in counts)  # Still False


# ============================================================
# 24. collections.defaultdict
# ============================================================

from collections import defaultdict

# defaultdict automatically creates a default value for missing keys.
default_counts = defaultdict(int)

for item in ["a", "b", "a", "c", "a"]:
    default_counts[item] += 1

print(dict(default_counts))

# defaultdict(list) is useful for grouping.
students_by_class = defaultdict(list)

students_by_class["Python"].append("Ada")
students_by_class["Python"].append("Grace")
students_by_class["Mathematics"].append("Alan")

print(dict(students_by_class))


# ============================================================
# 25. collections.Counter
# ============================================================

from collections import Counter

counter = Counter(["red", "blue", "red", "green", "red"])

print(counter)
print(counter["red"])
print(counter.most_common(2))


# ============================================================
# 26. JSON-style dictionaries
# ============================================================

# JSON objects are commonly represented as Python dictionaries.
json_like_data = {
    "name": "Ada",
    "age": 36,
    "skills": ["Python", "Mathematics"],
    "address": {
        "city": "London",
        "country": "UK",
    },
}

import json

# Convert a dictionary to a JSON string.
json_text = json.dumps(json_like_data, indent=2)
print(json_text)

# Convert a JSON string back into a dictionary.
restored_dictionary = json.loads(json_text)
print(restored_dictionary)


# ============================================================
# 27. Important dictionary rules
# ============================================================

rules = [
    "Dictionaries store key-value pairs.",
    "Keys must be unique.",
    "Keys must be hashable.",
    "Values may be any Python object.",
    "Dictionaries preserve insertion order.",
    "Direct access with [] raises KeyError for missing keys.",
    "get() provides safe access.",
    "Views returned by keys(), values(), and items() are dynamic.",
    "update() adds or replaces multiple items.",
    "copy() creates a shallow copy.",
    "Dictionaries are mutable.",
]

for number, rule in enumerate(rules, start=1):
    print(f"{number}. {rule}")


# ============================================================
# 28. Final practical example
# ============================================================

students = {
    "Ada": {"math": 95, "science": 90},
    "Grace": {"math": 88, "science": 94},
    "Alan": {"math": 91, "science": 86},
}

# Calculate each student's average.
averages = {
    name: sum(subjects.values()) / len(subjects)
    for name, subjects in students.items()
}

# Find the student with the highest average.
top_student = max(averages, key=averages.get)

print("Averages:", averages)
print("Top student:", top_student)
print("Top average:", averages[top_student])
```
