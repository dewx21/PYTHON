```python
# Python Lists: Beginner Tutorial

# A list stores multiple values in one variable.
fruits = ["apple", "banana", "orange"]
print("Original list:", fruits)

# Lists can contain different data types.
mixed_list = ["Python", 10, 3.14, True]
print("Mixed list:", mixed_list)

# Access items using indexes. Indexing starts at 0.
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])

# Change an item.
fruits[1] = "mango"
print("After changing an item:", fruits)

# Add one item to the end.
fruits.append("grape")
print("After append:", fruits)

# Insert an item at a specific position.
fruits.insert(1, "banana")
print("After insert:", fruits)

# Add multiple items to the end.
fruits.extend(["kiwi", "melon"])
print("After extend:", fruits)

# Combine two lists using +.
vegetables = ["carrot", "potato"]
food = fruits + vegetables
print("Combined lists:", food)

# Repeat a list using *.
repeated_list = ["A", "B"] * 3
print("Repeated list:", repeated_list)

# Python does not support direct list subtraction.
# Use a list comprehension to remove matching values.
numbers = [1, 2, 3, 4, 5]
values_to_remove = [2, 4]
remaining = [number for number in numbers if number not in values_to_remove]
print("After subtraction-like operation:", remaining)

# Remove an item by value.
fruits.remove("mango")
print("After remove:", fruits)

# Remove and return the item at an index.
last_fruit = fruits.pop()
print("Popped item:", last_fruit)
print("After pop:", fruits)

# Remove an item by index.
del fruits[0]
print("After del:", fruits)

# Find the position of an item.
print("Position of banana:", fruits.index("banana"))

# Count how many times an item appears.
fruits.append("banana")
print("Number of bananas:", fruits.count("banana"))

# Sort a list in ascending order.
numbers.sort()
print("Sorted numbers:", numbers)

# Sort in descending order.
numbers.sort(reverse=True)
print("Descending numbers:", numbers)

# Reverse the order of items.
fruits.reverse()
print("Reversed fruits:", fruits)

# Find the length of a list.
print("Number of fruits:", len(fruits))

# Check whether an item exists.
print("Is apple in the list?", "apple" in fruits)

# Loop through a list.
for fruit in fruits:
    print("Fruit:", fruit)

# Create a new list using a list comprehension.
squares = [number ** 2 for number in range(1, 6)]
print("Squared numbers:", squares)

# Copy a list.
fruits_copy = fruits.copy()
print("Copied list:", fruits_copy)

# Remove all items from a list.
fruits_copy.clear()
print("Cleared list:", fruits_copy)
```
